## 5. ファイル入出力

### 説明

計算結果や観測データは **ファイルに書き込み、後で読み込む** ことで再利用できます。
ここでは、物理・天文学で特によく使われる2通りの方法を紹介します。

- **数値配列（numpy）** → `np.savetxt` / `np.genfromtxt`
- **表形式データ（pandas）** → CSV / DataFrame




### 5.1 numpy + genfromtxt（数値配列の入出力）

シミュレーション結果や観測ログなど、
「列名のない数値データ」を扱う典型的な例です。

#### 書き込み

```python
# 自由落下の例
t = np.linspace(0, 5, 50)
g = 9.8
x = 0.5 * g * t**2

data = np.column_stack([t, x])
np.savetxt("free_fall.txt", data)
```

#### 読み込み

```python
fig = plt.figure()
ax = fig.add_subplot(111)

data = np.genfromtxt("free_fall.txt")

time = data[:, 0]
position = data[:, 1]

ax.plot(time, position)
ax.set_xlabel("time [s]")
ax.set_ylabel("position [m]")

fig.savefig("free_fall_from_txt.png", dpi=150)
plt.show()
```

#### 演習
上記のコードを実行してください。

### 5.2 CSV + DataFrame（表データの入出力）

列名を持つデータや、後処理・解析を行いたい場合はこちらが便利です。

#### 書き込み

```python
df = pd.DataFrame({
    "time": t,
    "position": x
})

df.to_csv("free_fall.csv", index=False)
```

#### 読み込み

```python
fig = plt.figure()
ax = fig.add_subplot(111)

df2 = pd.read_csv("free_fall.csv")

ax.plot(df2["time"], df2["position"])
ax.set_xlabel("time [s]")
ax.set_ylabel("position [m]")

fig.savefig("free_fall_from_csv.png", dpi=150)
plt.show()
```

## 6. 綺麗な図をつくるには

### 説明

論文・レポートでは **見やすい図** が重要です。特に以下に気をつけましょう。

- 軸ラベル
- 単位
- フォントサイズ（発表の時は特に大きめが良いです）

以下はよくある設定の例です。
```python
import matplotlib.pyplot as plt
fnameforfig="sans-serif"
fsizeforfig=14
fsizeforlabel=16
plt.rcParams['font.family'] = fnameforfig
plt.rcParams['font.size'] = fsizeforfig
plt.rcParams['xtick.direction'] = 'in'
plt.rcParams['ytick.direction'] = 'in'
plt.rcParams["xtick.minor.visible"] = True
plt.rcParams["ytick.minor.visible"] = True
plt.rcParams['xtick.top'] = True
plt.rcParams['ytick.right'] = True
```
複数の線を描く場合、凡例をつける必要があります。`ax.legend(frameon=False)`を忘れずに。

```python
fig = plt.figure()
ax = fig.add_subplot(111)
x = np.linspace(0, 2*np.pi, 100)
ax.plot(x, np.sin(x), label="sin")
ax.plot(x, np.cos(x), label="cos")
ax.set_xlabel("x [rad]")
ax.set_ylabel("value")
ax.legend(frameon=False)

fig.savefig("sin_cos.png", dpi=150)
plt.show()
```
